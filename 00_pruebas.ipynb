```python
from markitdown import MarkItDown
from openai import OpenAI

# Modelo vision-language disponible en Docker Model Runner
# ai/qwen3-vl soporta OCR e interpretación de imágenes de documentos escaneados
LLM_MODEL = "ai/qwen3-vl:4B-UD-Q4_K_XL"

# Docker Desktop → Models → expuesto en http://localhost:12434
client = OpenAI(
    base_url="http://localhost:12434/engines/llama.cpp/v1/",
    api_key="docker",
)
print(f"Backend LLM: Docker Model Runner → {LLM_MODEL}")

# enable_plugins=True activa el plugin de PDF que:
#   1. Detecta si el PDF no tiene capa de texto (escaneado)
#   2. Renderiza cada página como imagen (requiere poppler: brew install poppler)
#   3. Envía cada imagen al LLM de visión para OCR
md = MarkItDown(
    enable_plugins=True,
    llm_client=client,
    llm_model=LLM_MODEL,
)

PDF_PATH = "Normativa2026/Proyecto-de-Ley-Transformacion-Digital-y-Audiovisual.pdf"
print(f"Procesando: {PDF_PATH}")

result = md.convert(PDF_PATH)
print(f"\n{'═'*70}")
print(f"Caracteres extraídos: {len(result.text_content):,}")
print(f"{'═'*70}\n")
print(result.text_content[:3000])
```

```python
# Dependencias para procesar PDFs escaneados con visión LLM
# - markitdown[all]  → conversión de documentos + plugins PDF
# - pdf2image        → renderiza páginas PDF como imágenes (necesita poppler)
# - openai           → cliente OpenAI-compatible para Docker Model Runner
#
# Requisito del sistema (instalar una sola vez):
#   brew install poppler
import subprocess, sys

pkgs = ["markitdown[all]", "pdf2image", "openai"]
subprocess.run([sys.executable, "-m", "pip", "install", "-q", *pkgs], check=True)
print("Dependencias OK")
```
